```python
from transformers import pipeline
```

Using the "sentiment-analysis" capability in the pipeline

```python
classifier = pipeline('sentiment-analysis')
classifier('This is the best day in my life.')
```

```python
classifier('Oh my god, I need a first-aid right now!')
```

Using "zero-shot-classification" - classifies the data into different domains

```python
classifier=pipeline('zero-shot-classification')
classifier(
    'The Legislative Assembly is scheduled on 25th July on state-welfare',
    candidate_labels=['education','politics','business']
)
```

Using 'text-generation' - generates text based on the data we provide

```python

summarizer = pipeline(
    "summarization",
    model="ARTeLab/it5-summarization-fanpage-64",
    tokenizer="ARTeLab/it5-summarization-fanpage-64"
)

text = """
Blinkit is a leading quick-commerce platform in India that specializes in delivering groceries and everyday essentials to customers within minutes. The company was founded in 2013 as Grofers and initially operated as an online grocery delivery service. In 2021, it rebranded to Blinkit to emphasize its focus on ultra-fast deliveries. The platform aims to make shopping more convenient by providing instant access to a wide range of products through a mobile app and website.

Blinkit offers products across various categories, including fresh fruits and vegetables, dairy items, snacks, beverages, personal care products, household supplies, and electronics. To achieve rapid deliveries, the company operates a network of strategically located dark stores, which are small fulfillment centers designed to store inventory close to customers. This infrastructure helps reduce delivery times and ensures efficient order processing.

In 2022, Blinkit was acquired by Zomato, one of India's largest food delivery and restaurant discovery platforms. The acquisition strengthened Blinkit's market position and enabled it to expand its services to more cities. By combining advanced technology, data-driven logistics, and a customer-centric approach, Blinkit has become a significant player in India's growing quick-commerce industry.

Today, Blinkit continues to transform the way people shop for daily necessities by offering speed, convenience, and reliability. Its innovative business model has influenced the rapid growth of the quick-commerce sector, making it easier for customers to receive essential products at their doorstep in a matter of minutes.
"""

result = summarizer(
    text,
    max_length=120,
    min_length=40,
    do_sample=False,
    truncation=True
)

print(result[0]["summary_text"])
```
