```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('/content/drive/MyDrive/PLACEMENT AI/Dataset/Market_Basket_Optimisation.csv',header=None)
data
```

```python
data.isnull().sum()
```

```python
'''tra=[]
for i in range(len(data)):
  for j in range (20):
    if data.values[i,j]!='nan':
      tra.append(str(data.values[i,j]))'''
transactions = []
for i in range(len(data)):
    row = []
    for j in range(20):
        val = data.values[i, j]

        if str(val) != 'nan':
            row.append(str(val))

    transactions.append(row)


```

```python
transactions
```

```python
!pip install apyori
```

```python
from apyori import apriori
rules=apriori(transactions=transactions,min_support=.003,min_confidence=.2,min_length=2,max_length=2)
result=list(rules)
result
```

```python
results_list = []

for item in result:
    # item[1] is the support for the itemset
    # item[2] contains the rules (ordered_statistics)
    for rule in item[2]:
        # Extracting details from the OrderedStatistic object
        # rule[0] = items_base (Antecedent), rule[1] = items_add (Consequent)
        antecedent = ', '.join(list(rule[0]))
        consequent = ', '.join(list(rule[1]))
        support = item[1]
        confidence = rule[2]
        lift = rule[3]

        results_list.append([
            antecedent if antecedent else "(none)",
            consequent,
            support,
            confidence,
            lift
        ])

results_df = pd.DataFrame(
    results_list,
    columns=['Antecedent', 'Consequent', 'Support', 'Confidence', 'Lift']
)

results_df
```
