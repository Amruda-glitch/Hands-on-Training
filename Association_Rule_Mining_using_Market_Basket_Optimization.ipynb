```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/drive/MyDrive/Market_Basket_Optimisation.csv')
```

```python
transactions=[]
for i in range(0,len(df)):
    transactions.append([str(df.values[i,j]) for j in range(0,20) if str(df.values[i,j])!='nan'])
transactions
```

```python
min_support = 2
tid_sets = {}

for tid, transaction in enumerate(transactions, start=1):
    for item in transaction:
        if item not in tid_sets:
            tid_sets[item] = set()
        tid_sets[item].add(tid)
```

```python
print("Vertical Format:")
for item, tids in tid_sets.items():
    print(item, ":", tids)
```

```python
print("\nFrequent Itemsets:")
for item, tids in tid_sets.items():
    if len(tids) >= min_support:
        print((item,), "Support =", len(tids))
```

```python
from itertools import combinations
items = list(tid_sets.keys())

for item1, item2 in combinations(items, 2):
    common_tids = tid_sets[item1] & tid_sets[item2]

    if len(common_tids) >= min_support:
        print((item1, item2), "Support =", len(common_tids))
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth
import pandas as pd
```

```python
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df = pd.DataFrame(te_ary, columns=te.columns_)
```
