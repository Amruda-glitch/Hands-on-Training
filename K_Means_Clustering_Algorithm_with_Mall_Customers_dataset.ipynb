```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Genre']=le.fit_transform(df['Genre'])
df
```

```python
c=df.iloc[:,3:].values
c
```

```python
from sklearn.cluster import KMeans
```

```python
cs=[]
for i in range(1,11):
  kmeans=KMeans(n_clusters=i,init='k-means++',max_iter=300,n_init=10,random_state=0)
  kmeans.fit(c)
  cs.append(kmeans.inertia_)
plt.plot(cs)
```

```python
kmeans=KMeans(n_clusters=6,init=('k-means++'),random_state=42)
y_kmeans=kmeans.fit_predict(c)
```

```python
kmeans.inertia_
```

```python
kmeans.labels_
```

```python
plt.scatter(c[y_kmeans==0,0],c[y_kmeans==0,1],color='green',label='cluster1')
plt.scatter(c[y_kmeans==1,0],c[y_kmeans==1,1],color='blue',label='cluster2')
plt.scatter(c[y_kmeans==2,0],c[y_kmeans==2,1],color='orange',label='cluster3')
plt.scatter(c[y_kmeans==3,0],c[y_kmeans==3,1],color='black',label='cluster4')
plt.scatter(c[y_kmeans==4,0],c[y_kmeans==4,1],color='yellow',label='cluster5')
plt.scatter(c[y_kmeans==5,0],c[y_kmeans==5,1],color='purple',label='cluster6')
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],color='red',label='centroids',s=300,marker=',')
```
